#DEFINIÇÃO

##A convolução é uma operação matemática realizadas entre dois sinais gerando um terceiro sinal. para que a convolução seja possível o sistema envolvido, $x_i(t)$ e $y_i(t)$, deve ser linear e invariante ao tempo (LTI).

##1. Linearidade
<br>

$$
\Large
1.\quad T\{\,a_1 \cdot x_1(t) \;+\; a_2 \cdot x_2(t)\,\}\;=\; a_1 \cdot T\{x_1(t)\} \;+\; a_2 \cdot T\{x_2(t)\}
$$

<br>


##2. Invariância no Tempo
<br>

$$
\Large
2.\quad T\{\ x(t-t_0)\} \;=\; y(t-t_0)
$$

<br>

##A operação da convolução é representado pela expressão:
<br>

$$
\Large
y(t) = x(t) * h(t)
$$

<br>

##A operação da convolução consiste em inverter um dos sinais. Em seguida, multiplicá-los a cada iteração no domínio respectivo. Isso significa que a cada iteração um terceiro sinal é construido







#CONVOLUÇÃO NO DOMÍNIO DO TEMPO CONTÍNUO (t)







##No tempo continuo, um sistema LIT, que recebe sinais com domínios bem definidos, possui intervalos dentro do seu domínio infinitos pontos. A definição matemática para a convolução no domínio no tempo contínuo é expressa pela seguinte expressão:


<br>

$$
\Large
y(t) = x(t) * h(t) =  ∫_{-∞}^{∞} x(τ) \, \cdot \, h(t - τ) \, dτ
$$

<br>

##A resolução da integral consiste em inverter o sinal de h(t) e convoluir com o sinal x(t) de forma que o sinal h(t) passe por todo o sinal x(τ). esse processo ocorre da seguinte forma:

#CONVOLUÇÃO DOMÍNIO DO TEMPO DISCRETO (N)

##Computacionalmente, não é viável armazenar sinais no tempo contínuo, pois seria necessária memória infinita. Dessa forma, deve-se amostrar os sinais para obter um número finito de pontos e assim aplicar a operação de convolução. No entanto, como o domínio se torna discreto, a sua representação matemática é um pouco diferente:

$$
\Large
y[n] = x[n] * h[n] =  \sum_{k= 0 }^{N-1} x[k] \, \cdot \, h[n-k]
$$




##Portanto, a convolução de dois sinais ocorrerá da forma apresentada anteriormente. Suponha-se, então, que se tenham os seguintes sinais:

$
\Large
h[n] = [1,0]
$

$
\Large
x[n] = [2,1,3]
$
<br>
$$
\Large
y[n] = x[n] * h[n] =  \sum_{k= 0 }^{ 3} x[k] \, \cdot \, h[n-k]
$$
<br>
$
\Large
y(0) = x(0) \cdot h(0) \,\, + \,\, x(1) \cdot h(-1) \,\, + \,\, x(2) \cdot h(-2) = 2
$

$
\Large
y(1) = x(0) \cdot h(1) \,\, + \,\, x(1) \cdot h(0) \,\, + \,\, x(2) \cdot h(-1) = 1
$

$
\Large
y(2) = x(0) \cdot h(2) \,\, + \,\, x(1) \cdot h(1) \,\, + \,\, x(2) \cdot h(0) = 3
$

$
\Large
y(0) = x(0) \cdot h(3) \,\, + \,\, x(1) \cdot h(2) \,\, + \,\, x(2) \cdot h(1) = 0
$

##resultado:

$
\Large
y = [2,1,3,0]
$



#CONVOLUÇÃO VETORIZADA

##Sabendo que a implementação computacional da convolução é expressa por

$$
\Large
y[n] = x[n] * h[n] =  \sum_{k= 0 }^{N-1} x[k] \, \cdot \, h[n-k]
$$

##Pode-se obter o resultado da convolução através do produto escalar entre a matriz de resposta ao impulso e o vetor do sinal de entrada. Então tendo os seguintes dados:
<br>
<div align="center">
<font size="3.5">
$h = \begin{bmatrix} h_0 \\ h_1 \\ h_2 \\ \vdots \\ h_{M-1} \end{bmatrix}_{M \times 1}\quad \quad x = \begin{bmatrix} x_0 \\ x_1 \\ x_2 \\ \vdots \\ x_{N-1} \end{bmatrix}_{N \times 1} \quad \quad H = \begin{bmatrix} h_{0} & h_{-1} & h_{-2} & \dots & h_{1-k} \\ h_{1} & h_{0} & h_{-1} & \dots & h_{2-k} & \\ h_{2} & h_{1} & h_{0} & \dots & h_{3-k} \\ \vdots & \vdots & \vdots & \ddots & \vdots \\ h_{n} & h_{n-1} & h_{n-2} & \dots & h_{n- k} \end{bmatrix}_{(M + N -1) \times M }$
</font>
</div>
<br>

##Através desses dados obtem-se a convolução:
<br>

<div align="center">
<font size="4">
$y = \begin{bmatrix} h_{0} & h_{-1} & h_{-2} & \dots & h_{1-k} \\ h_{1} & h_{0} & h_{-1} & \dots & h_{2-k} \\ h_{2} & h_{1} & h_{0} & \dots & h_{3-k}  \\ \vdots & \vdots & \vdots & \ddots & \vdots \\ h_{n-1} & h_{n-2} & h_{n-3} & \dots & h_{n-k} \end{bmatrix} \cdot \, \begin{bmatrix} x_1 \\ x_2 \\ x_3 \\ \vdots \\ x_{N-1} \end{bmatrix}$
</font>
</div>
<br>

##onde y é um vetor de dados:

<br>

<div align="center">
<font size="3">
$y = \begin{bmatrix} y_1 \\ y_2 \\ y_3 \\ \vdots \\ y_{n} \end{bmatrix}_{M + N -1}$
</font>
</div>

##A matriz de resposta aoa impulso é obtida tomando como base a matriz de teodiplitz. observe o exemplo a seguir:

<div align="center">
<font size="4">
$
A = \begin{bmatrix}
1 & 2 & 3 \\
4 & 1 & 2 \\
5 & 4 & 1 \\
6 & 5 & 4
\end{bmatrix}_{4 \times 3}
$
</font>
</div>


##Para um H (matriz referente a h) de mesma dimensão, por exemplo:
<br>
<div align="center">
<font size="5">
$
h = [h_{0} \,\, h_{1} \,\, h_{2}]_{1 × 3}
$
</font>
</div>
<br><br>

<div align="center">
<font size="5">
$
H = \begin{bmatrix}
h_0 & h_{-1} & h_{-2} \\
h_1 & h_0 & h_{-1} \\
h_2 & h_1 & h_0 \\
h_3 & h_2 & h_1
\end{bmatrix}_{4 \times 3}
$
</font>
</div>
<br>

##onde tem-se a representação genérica:

<br>
<div align="center">
<font size="5">
$H = \begin{bmatrix} h_{0} & h_{-1} & h_{-2} & \dots & h_{1-k} \\ h_{1} & h_{0} & h_{-1} & \dots & h_{2-k} & \\ h_{2} & h_{1} & h_{0} & \dots & h_{3-k} \\ \vdots & \vdots & \vdots & \ddots & \vdots \\ h_{n} & h_{n-1} & h_{n-2} & \dots & h_{n- k} \end{bmatrix}_{(M + N -1) \times M }$
</font>
</div>

```python
import numpy as np
import matplotlib.pyplot as plt
import time, tracemalloc
from IPython.display import display, Math
import sympy as sp
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

# FUNÇÕES

```python
def convolucao_linear(x,h): #aplica a forma analitica da convolução no tempo discreto

  p = 0 #variável utilizada para acumular os valores do somatório, determinando o valor da convolução na posição n

  y = [] #vetor que armazena o resulado  da convolução

  for n in range(len(x)+len(h)-1): #percorre os valores de n

    for k in range(len(x)): #percorre os valores de k

      p+=x[k]*h[n-k] if (n-k) >=0 and (n-k)< len(h) else 0 #determina o valor da convolução no instante n. Caso (n-k) seja um indice pertecente ao vetor h calcula o valor, caso contrário o resultado é 0

    y.append(p) #após calcular o somatório para um derteminado valor de n armazena no vetor resultado da convolução

    p = 0 #reinicia o acumulador para a proxima iteração n

  return y #retorna o resultado da convolução


def convolucao_linear_vet(x,h): #aplica a forma vetorizada da convolução no tempo discreto

#OBS: #a forma vetorizada aplica a convolução denominada circular, como é necessario a convolução linear para obter o resultado esperado é necessário fazer zero padding, tornando o tamanho de h e x iguais

  if len(x) < len(h): #se x menor que h será aplicado zero padding em x para ter o mesmo tamanho de h

    x = np.pad(x, (0, len(h)-len(x)), 'constant') #aplica o zero padding

  h = np.pad(h, (0, len(x)-len(h)), 'constant')# caso contrário(len(x) => len(h)) aplica o zero padding em h

  D = np.arange(len(h) + len(x) - 1).reshape(-1,1) - np.arange(len(h)) #calcula a Matriz de diferença

  '''  print("\nMatriz de diferença(Indices da resposta ao impulso):\n")
    display(Math("D_{i,j} =  h^T - h = " + sp.latex(sp.Matrix(D))))'''


  H = np.append(h,np.zeros(len(h)-1,dtype=int))[D] #Matriz da resposta ao impulso obtida a apartir da matriz de diferença

  '''  print("\nIndexação para associar os indices aos valores de h: \n")
    display(Math("H = " + sp.latex(sp.Matrix(H))))'''

  return H @ x #Resultado da convolução vetorizada



def grafico_plotly(x, y, h):
  # Crie uma figura com 3 subplots
  fig = make_subplots(
      rows=1, cols=3,
      subplot_titles=("Sinal de entrada x[n]", "Resposta ao impulso h[n]", "Convolução y[n]")
  )

  # --- Sinal de entrada (x) ---
  fig.add_trace(go.Bar(
      x=np.arange(len(x)),
      y=x,
      marker=dict(color='blue'), # <-- 'width' foi removido daqui
      width=0.4, # <-- 'width' foi adicionado aqui
      name='Sinal x[n]',
      showlegend=False
  ), row=1, col=1)

  fig.add_trace(go.Scatter(
      x=np.arange(len(x)),
      y=x,
      mode='markers',
      marker=dict(color='blue', size=8),
      name='Sinal x[n]',
      showlegend=False
  ), row=1, col=1)


  # --- Resposta ao impulso (h) ---
  fig.add_trace(go.Bar(
      x=np.arange(len(h)),
      y=h,
      marker=dict(color='red'), # <-- 'width' foi removido daqui
      width=0.4, # <-- 'width' foi adicionado aqui
      name='Resposta h[n]',
      showlegend=False
  ), row=1, col=2)

  fig.add_trace(go.Scatter(
      x=np.arange(len(h)),
      y=h,
      mode='markers',
      marker=dict(color='red', size=8),
      name='Resposta h[n]',
      showlegend=False
  ), row=1, col=2)


  # --- Convolução (y) ---
  fig.add_trace(go.Bar(
      x=np.arange(len(y)),
      y=y,
      marker=dict(color='green'), # <-- 'width' foi removido daqui
      width=0.4, # <-- 'width' foi adicionado aqui
      name='Convolução y[n]',
      showlegend=False
  ), row=1, col=3)

  fig.add_trace(go.Scatter(
      x=np.arange(len(y)),
      y=y,
      mode='markers',
      marker=dict(color='green', size=8),
      name='Convolução y[n]',
      showlegend=False
  ), row=1, col=3)

  # --- Ajustes de Layout e Eixos ---
  fig.update_layout(
      title_text="Convolução: Sinal de Entrada, Resposta ao Impulso e Saída",
      height=500,
      showlegend=True
  )

  fig.update_yaxes(range=[0, 7], row=1, col=1)
  fig.update_yaxes(range=[0, 7], row=1, col=2)
  fig.update_yaxes(range=[0, 7], row=1, col=3)

  fig.show()
```

```python
import time
import numpy as np
import matplotlib.pyplot as plt
import tracemalloc

# --- Funções de Convolução ---

def conv_laco(x, h):
    """(1) Convolução com laços for (versão 1)."""
    N = len(x); M = len(h)
    y = [0.0] * (N + M - 1)
    for i in range(N):
        for j in range(M):
            y[i + j] += x[i] * h[j]
    return y

def convolucao_soma_direta(x,h):
    """(2) Convolução por somatório direto (sua nova versão)."""
    p = 0
    y = []
    for n in range(len(x)+len(h)-1):
        for k in range(len(x)):
            p+=x[k]*h[n-k] if (n-k) >=0 and (n-k)< len(h) else 0
        y.append(p)
        p = 0
    return y

def convolucao_linear_vet(x,h):
    """(3) Convolução via Matriz de Toeplitz."""
    x = np.asarray(x); h = np.asarray(h)
    if len(x) < len(h):
        x = np.pad(x, (0, len(h)-len(x)), 'constant')
    elif len(h) < len(x):
        h = np.pad(h, (0, len(x)-len(h)), 'constant')

    L = len(h)
    D = np.arange(L + len(x) - 1).reshape(-1,1) - np.arange(L)
    H = np.append(h,np.zeros(L-1,dtype=int))[D]
    return H @ x

def conv_vetorizada_outer(x, h):
    """(4) Convolução via produto externo (outer product)."""
    comb_x_h = np.outer(x, h)
    y = []
    for k in range(-comb_x_h.shape[0] + 1, comb_x_h.shape[1]):
        y.append(np.sum(np.flipud(comb_x_h).diagonal(k)))
    return np.array(y)

def conv_numpy_optimized(x, h):
    """(5) Convolução usando a função otimizada np.convolve."""
    return np.convolve(x, h)

# --- Configuração do Experimento ---
NUM_EXECUCOES = 5
TAMANHO_KERNEL_FIXO = 150
tamanhos_sinal = np.arange(200, 3000, 250)

resultados = {
    'laco': {'tempos': [], 'memorias': []},
    'soma_direta': {'tempos': [], 'memorias': []},
    'toeplitz': {'tempos': [], 'memorias': []},
    'outer': {'tempos': [], 'memorias': []},
    'numpy_opt': {'tempos': [], 'memorias': []}
}

print(f"Iniciando medição com 5 algoritmos (média de {NUM_EXECUCOES} execuções)...")

for tamanho in tamanhos_sinal:
    print(f"Testando com sinal de tamanho: {tamanho}")
    np.random.seed(42)
    sinal_x = np.random.rand(tamanho)
    kernel_h = np.random.rand(TAMANHO_KERNEL_FIXO)
    sinal_x_lista = list(sinal_x)
    kernel_h_lista = list(kernel_h)

    temp_res = {key: {'t': [], 'm': []} for key in resultados}

    for _ in range(NUM_EXECUCOES):
        # 1. Laço For
        tracemalloc.start()
        start_time = time.perf_counter()
        conv_laco(sinal_x_lista, kernel_h_lista)
        _, peak_mem = tracemalloc.get_traced_memory()
        tracemalloc.stop()
        temp_res['laco']['t'].append(time.perf_counter() - start_time)
        temp_res['laco']['m'].append(peak_mem)

        # 2. Soma Direta
        tracemalloc.start()
        start_time = time.perf_counter()
        convolucao_soma_direta(sinal_x_lista, kernel_h_lista)
        _, peak_mem = tracemalloc.get_traced_memory()
        tracemalloc.stop()
        temp_res['soma_direta']['t'].append(time.perf_counter() - start_time)
        temp_res['soma_direta']['m'].append(peak_mem)

        # 3. Matriz Toeplitz
        tracemalloc.start()
        start_time = time.perf_counter()
        convolucao_linear_vet(sinal_x, kernel_h)
        _, peak_mem = tracemalloc.get_traced_memory()
        tracemalloc.stop()
        temp_res['toeplitz']['t'].append(time.perf_counter() - start_time)
        temp_res['toeplitz']['m'].append(peak_mem)

        # 4. Outer Product
        tracemalloc.start()
        start_time = time.perf_counter()
        conv_vetorizada_outer(sinal_x, kernel_h)
        _, peak_mem = tracemalloc.get_traced_memory()
        tracemalloc.stop()
        temp_res['outer']['t'].append(time.perf_counter() - start_time)
        temp_res['outer']['m'].append(peak_mem)

        # 5. NumPy Otimizado
        tracemalloc.start()
        start_time = time.perf_counter()
        conv_numpy_optimized(sinal_x, kernel_h)
        _, peak_mem = tracemalloc.get_traced_memory()
        tracemalloc.stop()
        temp_res['numpy_opt']['t'].append(time.perf_counter() - start_time)
        temp_res['numpy_opt']['m'].append(peak_mem)

    for key in resultados:
        resultados[key]['tempos'].append(np.mean(temp_res[key]['t']))
        resultados[key]['memorias'].append(np.mean(temp_res[key]['m']))

print("Medição concluída.")

# --- Geração dos Gráficos ---

legend_names = {
    'laco': '1. Laço For',
    'soma_direta': '2. Soma Direta',
    'toeplitz': '3. Matriz Toeplitz',
    'outer': '4. Vetorizado (Outer Product)',
    'numpy_opt': '5. NumPy (np.convolve)'
}
markers = {'laco': 'o', 'soma_direta': 'v', 'toeplitz': 'D', 'outer': '^', 'numpy_opt': 's'}

# Gráfico 1: Tempo
plt.figure(figsize=(12, 8))
for key, data in resultados.items():
    plt.plot(tamanhos_sinal, data['tempos'], label=legend_names[key], marker=markers[key])
plt.xlabel('Tamanho do Sinal de Entrada'); plt.ylabel('Tempo de Execução (segundos)')
plt.title(f'Tempo de Execução (Média de {NUM_EXECUCOES} Execuções)'); plt.legend()
plt.grid(True, which="both", ls="--"); plt.yscale('log'); plt.tight_layout()
plt.savefig('comparacao_tempo_5_metodos.png')
print("\nGráfico 'comparacao_tempo_5_metodos.png' foi gerado.")

# Gráfico 2: Memória
plt.figure(figsize=(12, 8))
for key, data in resultados.items():
    mem_kb = np.array(data['memorias']) / 1024
    plt.plot(tamanhos_sinal, mem_kb, label=legend_names[key], marker=markers[key])
plt.xlabel('Tamanho do Sinal de Entrada'); plt.ylabel('Pico de Memória RAM (KB)')
plt.title(f'Consumo de Memória (Média de {NUM_EXECUCOES} Execuções)'); plt.legend()
plt.grid(True, which="both", ls="--"); plt.tight_layout()
plt.savefig('comparacao_memoria_5_metodos.png')
print("Gráfico 'comparacao_memoria_5_metodos.png' foi gerado.")
```
