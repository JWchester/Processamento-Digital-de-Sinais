# Bibliotecas


```python
import numpy as np
import plotly.graph_objects as go
import matplotlib.pyplot as plt
import time
import math
import tracemalloc
```

# Funções

```python
def convolucao_linear(x,h):
  p = 0
  y = []
  for n in range(len(x)+len(h)-1):
    for k in range(len(x)):
      p+=x[k]*h[n-k] if (n-k) >=0 and (n-k)< len(h) else 0

    y.append(p)
    p = 0
  return y


def convolucao_por_partes(x,h,L):
  x = x + [0] * (math.ceil(len(x)/L)*L - len(x))
  y = np.zeros(len(x) +len(h) -1)

  for i in range(0,len(y),L):
    y[i:L+len(h)-1+i] += convolucao_linear(x[i:L+i],h)
  return y
```

# Implementação

```python
tracemalloc.start() # inicia o processo para calcular o consumo de memoria RAM
start = time.perf_counter() # inicia o processo para calcular o tempo de execução do algoritmo

x = [1, -1, 1, -1, 1]
h = [1, 1, 1]
L = 2

y = convolucao_por_partes(x,h,L)
print(y)

end = time.perf_counter() # termina o processo para calcular o tempo de execução do algoritmo
current, peak = tracemalloc.get_traced_memory() # armazena os valores de memoria RAM final e de pico
tracemalloc.stop() # termina o processo para calcular o consumo de memória RAM

# imprime os resultados do tempo de execução e consumo de memoria
print(f"Tempo: {end - start:.6f} segundos")
print(f"Memória atual: {current / 1024:.2f} KB")
print(f"Pico de memória: {peak / 1024:.2f} KB")
```
